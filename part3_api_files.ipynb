```python
# Python script to write, append, read, and search in python_notes.txt

file_path = "python_notes.txt"

# Part A — Write initial notes
initial_lines = [
    "Topic 1: Variables store data. Python is dynamically typed.",
    "Topic 2: Lists are ordered and mutable.",
    "Topic 3: Dictionaries store key-value pairs.",
    "Topic 4: Loops automate repetitive tasks.",
    "Topic 5: Exception handling prevents crashes.",
]

with open(file_path, "w", encoding="utf-8") as f:
    for line in initial_lines:
        f.write(line + "\n")
print("File written successfully.")

# Part A — Append two more lines
append_lines = [
    "Topic 6: Functions encapsulate reusable logic.",
    "Topic 7: Modules help organize code across files."
]

with open(file_path, "a", encoding="utf-8") as f:
    for line in append_lines:
        f.write(line + "\n")
print("Lines appended.")

# Part B — Read and print numbered lines
with open(file_path, "r", encoding="utf-8") as f:
    lines = [line.rstrip("\n") for line in f]

for idx, line in enumerate(lines, start=1):
    print(f"{idx}. {line}")

total_lines = len(lines)
print(f"Total number of lines: {total_lines}")

# Ask user for keyword and print matching lines (case-insensitive)
keyword = input("Enter a keyword to search for: ").strip()
if keyword == "":
    print("No keyword entered.")
else:
    lower_kw = keyword.lower()
    matches = [(idx + 1, line) for idx, line in enumerate(lines) if lower_kw in line.lower()]
    if matches:
        print(f"Lines containing '{keyword}':")
        for idx, line in matches:
            print(f"{idx}. {line}")
    else:
        print(f"No lines found containing '{keyword}'.")
```

```python
# Part A — Guarded Calculator
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Cannot divide by zero"
    except TypeError:
        return "Error: Invalid input types"

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide("ten", 2))

# Part B — Guarded File Reader
def read_file_safe(filename):
    try:
        with open(filename, "r", encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return None
    finally:
        print("File operation attempt complete.")

content1 = read_file_safe("python_notes.txt")
if content1 is not None:
    print("python_notes.txt content:")
    print(content1)

content2 = read_file_safe("ghost_file.txt")
if content2 is not None:
    print("ghost_file.txt content:")
    print(content2)

# Part C — Robust API Calls
import requests  # already imported in the notebook, but safe to ensure availability

def safe_get(url, timeout=5):
    try:
        return requests.get(url, timeout=timeout)
    except requests.exceptions.ConnectionError:
        print("Connection failed. Please check your internet.")
        return None
    except requests.exceptions.Timeout:
        print("Request timed out. Try again later.")
        return None
    except Exception as e:
        print(f"Unexpected error: {e}")
        return None

def safe_post(url, json=None, timeout=5):
    try:
        return requests.post(url, json=json, timeout=timeout)
    except requests.exceptions.ConnectionError:
        print("Connection failed. Please check your internet.")
        return None
    except requests.exceptions.Timeout:
        print("Request timed out. Try again later.")
        return None
    except Exception as e:
        print(f"Unexpected error: {e}")
        return None

url_all = "https://dummyjson.com/products"
url_laptops = "https://dummyjson.com/products/category/laptops"
post_url = "https://dummyjson.com/products/add"

print("=" * 100)
print("STEP 1: Fetch and Display Products")
print("=" * 100)
resp = safe_get(url_all)
if resp is not None:
    if resp.status_code == 200:
        data = resp.json()
        products = data.get('products', [])
        print(f"\n{'ID':<5} | {'Title':<30} | {'Category':<15} | {'Price':<10} | {'Rating':<7}")
        print("-" * 100)
        for product in products:
            pid = product.get('id')
            title = str(product.get('title', '')).ljust(30)
            category = str(product.get('category', '')).ljust(15)
            price = f"${product.get('price', '')}"
            rating = product.get('rating', '')
            print(f"{pid:<5} | {title} | {category} | {price:<10} | {rating:<7}")
    else:
        print(f"GET {url_all} returned status code {resp.status_code}")

print("\n" + "=" * 100)
print("STEP 2: Filter (rating ≥ 4.5) and Sort by Price (Descending)")
print("=" * 100)
if resp is not None and resp.status_code == 200:
    filtered = [p for p in products if p.get('rating', 0) >= 4.5]
    filtered_sorted = sorted(filtered, key=lambda x: x.get('price', 0), reverse=True)
    print(f"\n{'ID':<5} | {'Title':<30} | {'Category':<15} | {'Price':<10} | {'Rating':<7}")
    print("-" * 100)
    for product in filtered_sorted:
        pid = product.get('id')
        title = str(product.get('title', '')).ljust(30)
        category = str(product.get('category', '')).ljust(15)
        price = f"${product.get('price', '')}"
        rating = product.get('rating', '')
        print(f"{pid:<5} | {title} | {category} | {price:<10} | {rating:<7}")

print("\n" + "=" * 100)
print("STEP 3: Fetch and Display Laptops")
print("=" * 100)
resp2 = safe_get(url_laptops)
if resp2 is not None:
    if resp2.status_code == 200:
        laptops_data = resp2.json().get('products', [])
        print(f"\n{'Brand':<20} | {'Title':<40} | {'Price':<10}")
        print("-" * 80)
        for laptop in laptops_data:
            brand = str(laptop.get('brand', '')).ljust(20)
            title = str(laptop.get('title', '')).ljust(40)
            price = f"${laptop.get('price', '')}"
            print(f"{brand} | {title} | {price:<10}")
    else:
        print(f"GET {url_laptops} returned status code {resp2.status_code}")

print("\n" + "=" * 100)
print("STEP 4: POST Request - Create Custom Product")
print("=" * 100)
payload = {
    "title": "My Custom Product",
    "price": 999,
    "category": "electronics",
    "description": "A product I created via API"
}
resp3 = safe_post(post_url, json=payload)
if resp3 is not None:
    print(f"\nStatus Code: {resp3.status_code}")
    try:
        print("Response:")
        print(resp3.json())
    except Exception as e:
        print(f"Could not parse JSON response: {e}")

# Part D — Input Validation Loop
while True:
    user_input = input("Enter a product ID to look up (1–100), or 'quit' to exit: ").strip()
    if user_input.lower() == "quit":
        break
    if not user_input.isdigit():
        print("Invalid input. Please enter an integer between 1 and 100 or 'quit'.")
        continue
    prod_id = int(user_input)
    if not (1 <= prod_id <= 100):
        print("Invalid input. Please enter an integer between 1 and 100.")
        continue

    detail_url = f"https://dummyjson.com/products/{prod_id}"
    resp4 = safe_get(detail_url)
    if resp4 is None:
        continue
    if resp4.status_code == 404:
        print("Product not found.")
    elif resp4.status_code == 200:
        try:
            p = resp4.json()
            print(f"Title: {p.get('title', 'N/A')}, Price: {p.get('price', 'N/A')}")
        except Exception as e:
            print(f"Failed to parse product info: {e}")
    else:
        print(f"Received unexpected status code: {resp4.status_code}")
```
